Noticed that some players in the data frames have multiple records. The main reason is that those players were traded to another team during the season.

The duplication of players has to be re-moved. The last row of each duplicated player will remain and perfor-mance data will be replaced by the first row of the duplicated player.


## Data Cleaning

```python
# Extract player with '2TM', '3TM' and '4TM' in column 'Team'

df1_multi_team = df1[(df1['Team'] == '2TM') | (df1['Team'] == '3TM') | (df1['Team'] == '4TM')]
df2_multi_team = df2[(df2['Team'] == '2TM') | (df2['Team'] == '3TM') | (df2['Team'] == '4TM')]
df3_multi_team = df3[(df3['Team'] == '2TM') | (df3['Team'] == '3TM') | (df3['Team'] == '4TM')]
```

```python
def clean_nba_multi_team_records(df, df_multi_team, cols_to_drop=[]):
    """
    A dedicated function for processing player data from mid-season transfers.
    Overwrites a player's last team record with TOT (Total) total data and removes specified unnecessary fields.

    Parameters:
    - df: The original DataFrame containing all player season data (including individual team records)
    - df_tot: A DataFrame containing only the TOT (Total) data for transferred players
    - cols_to_drop: The list of feature fields to ultimately remove
    """
    
    df_working = df.copy()
    df_multi_team_working = df_multi_team.copy()
    
    # Create a new column 'ID' with unique numbers for each row
    df_working['ID'] = df_working.index + 1
    
    # Get the max ID of each player
    df_max_id_rows = df_working.loc[df_working.groupby('Player')['ID'].idxmax()].copy()
    df_max_id_rows = df_max_id_rows.sort_values(by='ID')
    
    # Set column 'Player' as the index of the dataframes
    df_max_id_rows.set_index('Player', inplace=True)
    df_multi_team_working.set_index('Player', inplace=True)
    
    # Update columns with performance data in df_max_id_rows using corresponding values from df_multi_team
    df_max_id_rows.update(df_multi_team_working)
    
    # Reset the index
    df_max_id_rows.reset_index(inplace=True)
    
    # Drop unnecessary columns
    cols_to_remove = cols_to_drop + ['ID']
    df_max_id_rows.drop(columns=cols_to_remove, axis=1, inplace=True, errors='ignore')
    
    return df_max_id_rows
```
